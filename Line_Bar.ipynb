```python
import matplotlib.pyplot as plt
```

```python
categories = ["A", "B", "C", "D"]
values = [10, 20, 15, 25]
x = [0, 1, 2, 3]
```

```python
#line plot
plt.plot(x, values, color = "blue", marker="o", label="Sales")
plt.title("Line plot")
plt.xlabel("categories")
plt.ylabel("values")

plt.xticks(x, categories)
plt.yticks(range(0, 36, 5))

plt.xlim(-0.5, 3.5)
plt.ylim(0, 35)

plt.grid(True)
plt.legend()
```

```python
#Bar plot
plt.bar(x, values, color="orange", label="sales")

plt.title("Bar chart")
plt.xlabel("categories")
plt.ylabel("values")

plt.xticks(x, categories)
plt.yticks(range(0, 36, 5))

plt.xlim(-0.5, 3.5)
plt.ylim(0, 35)

plt.grid(axis = "y")
plt.legend()

plt.text(3, 27, "Highest", ha="center")

plt.savefig("graph.png")
```
