```python
import pandas as pd
```

```python
# Load Dataset
df = pd.read_csv("Toyota.csv", na_values=["??", "????"])

df = df.drop(columns=["Unnamed: 0"])

df["Doors"] = df["Doors"].replace({
    "three": "3",
    "four": "4",
    "five": "5"
})

df["Doors"] = pd.to_numeric(df["Doors"])
```

```python
print("Original Dataset:")
display(df.head())

print("Shape:", df.shape)
print("Missing Values:")
print(df.isnull().sum())
```

```python
# 1. Drop Missing Values
df1 = df.dropna()

print("After Dropping Missing Values:")
display(df1.head())

# 2. Fill Missing Values with Zero
df2 = df.fillna(0)

print("After Filling with Zero:")
display(df2.head())

# 3. Fill Missing Values with Mean
df3 = df.copy()

num = df3.select_dtypes(include="number").columns
df3[num] = df3[num].fillna(df3[num].mean())

df3["FuelType"] = df3["FuelType"].fillna(df3["FuelType"].mode()[0])

print("After Filling with Mean:")
display(df3.head())

# 4. Handle Duplicate Rows
print("Duplicate Rows:", df3.duplicated().sum())

df4 = df3.drop_duplicates()

print("After Removing Duplicates:")
display(df4.head())
```

```python
# 5. Detect and Remove Outliers using Z-score
columns = ["Price", "Age", "KM", "HP", "CC", "Weight"]

z = ((df4[columns] - df4[columns].mean()) / df4[columns].std()).abs()

outliers = (z > 3).any(axis=1)

print("Number of Outliers:", outliers.sum())

df5 = df4[~outliers].copy()

print("After Removing Outliers:")
display(df5.head())

print("Final Shape:", df5.shape)
```
