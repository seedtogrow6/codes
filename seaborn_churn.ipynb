```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Orange_Telecom_Churn_Data.csv")
print("first 5 rows : ")
print(df.head())

print("dataset shape : ")
print(df.shape)

print("missing values : ")
print(df.isnull().sum())
```

```python
#histogram

sns.histplot(x = df["total_day_minutes"], bins = 20, kde = True)
plt.title("Histogram")
plt.xlabel("total day minutes")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.boxplot(y = df["total_night_minutes"])
plt.title("box plot")
plt.ylabel("Total night minutes")
plt.show()
```

```python
plt.figure(figsize = (10,10))
corr = df.corr(numeric_only = True)
sns.heatmap(corr, cmap = "coolwarm")
plt.title("correlation heatmap")
plt.tight_layout()
plt.show()
```
