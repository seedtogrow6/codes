```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("data.db")
cur = conn.cursor()
```

```python
cur.execute("""
        CREATE TABLE IF NOT EXISTS students (
            id INTEGER PRIMARY KEY,
            name TEXT,
            age INTEGER,
            grade TEXT
        )
    """)
```

```python
cur.execute("INSERT OR IGNORE INTO students VALUES (1, 'ABC', 22, 'A')")
cur.execute("INSERT OR IGNORE INTO students VALUES (2, 'PQR', 25, 'B')")
cur.execute("INSERT OR IGNORE INTO students VALUES (3, 'XYZ', 20, 'A')")
cur.execute("INSERT OR IGNORE INTO students VALUES (4, 'Neha', 23, 'B')")

conn.commit()
```

```python
# Import SQL table into DataFrame
df = pd.read_sql_query("SELECT * FROM students", conn)

print("Original DataFrame:")
display(df)

# 1. Display first 2 rows
print("First 2 Rows:")
display(df.head(2))

# 2. Filter students with age > 20
print("Students above age 20:")
display(df[df["age"] > 20])

# 3. Sort by age
print("Sorted by Age:")
display(df.sort_values(by="age"))

# 4. Calculate average age
print("Average Age:")
print(df["age"].mean())

# 5. Count students by grade
print("Grade Count:")
print(df["grade"].value_counts())

conn.close()
```
