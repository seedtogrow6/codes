```python
import requests
import pandas as pd
```

```python
API_KEY = "e79214a1d3ac5ef9ed1a98c59809a742"
url = "https://api.openweathermap.org/data/2.5/weather"
```

```python
cities = ["Pune", "Mumbai", "Delhi", "Chennai"]
records = []

for city in cities:
    params = {
        "q" : city,
        "appid" : API_KEY,
        "units" : "metric"
    }

response = requests.get(url, params = params)
```

```python
if response.status_code == 200:
    data = response.json()

    records.append({
        "City": data["name"],
        "Temperature" : data["main"]["temp"],
        "Humidity" : data["main"]["humidity"],
        "Wind" : data["wind"]["speed"]
    })
else:
    print("Error for", City, response.status_code)
```

```python
df = pd.DataFrame(records)

print("first 5 rows : ")
print(df.head())

print("Statistical summary : ")
print(df.describe())

print("sort by temperature : ")
print(df.sort_values("Temperature"))

print("cities with temperature above 30 : ")
print(df[df["Temperature"]>30])

print("Average Temperature : ")
print(df["Temperature"].mean())

df.to_csv("weather_data.csv", index = False)
```
