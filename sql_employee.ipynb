```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect("employee.db")
cur = conn.cursor()
```

```python
cur.execute("""
    CREATE TABLE IF NOT EXISTS employee(
        id INTEGER PRIMARY KEY,
        name TEXT,
        age INT,
        salary INT
    )
""")
```

```python
data = [
    (1, "rohit", 19, 10000),
    (2, "krushna", 20, 20000),
    (3, "dp", 20, 30000),
    (4, "yash", 24, 5000),
    (5, "prem", 18, 2000)
]

cur.executemany(
    "INSERT OR REPLACE INTO employee VALUES (?, ?, ?, ?)",
    data
)
```

```python
conn.commit()
```

```python
df = pd.read_sql_query("SELECT * FROM employee", conn)
display(df)
```

```python
#1 display
print(df.head(3))
```

```python
#2 filter by salary
print(df[df["salary"]>6000])
```

```python
#sort by salary
print(df.sort_values(by="salary", ascending=False))
```

```python
#mean
print(df["salary"].mean())
```
