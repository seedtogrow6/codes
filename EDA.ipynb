```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris = load_iris(as_frame = True)
df = iris.frame
```

```python
print("first 5 rows")
print(df.head())

print("last 5 Rows")
print(df.tail())
```

```python
print("Shape : ")
print(df.shape)
```

```python
print("dataset information : ")
print(df.info())
```

```python
print("Statistical summary : ")
print(df.describe())
```

```python
print("null values : ")
print(df.isnull().sum())
```

```python
print("Species count : ")
print(df["target"].value_counts())
```

```python
plt.bar(df["target"].value_counts().index,
        df["target"].value_counts().values,
        color="orange")

plt.xticks([0, 1, 2])

plt.xlabel("Species")
plt.ylabel("Count")
plt.title("Species Count")
plt.show()
```

```python
plt.hist(df["sepal length (cm)"], bins = 10)
plt.xlabel("sepel length")
plt.title("sepel length distribution")
plt.show()
```

```python
plt.scatter(df["petal length (cm)"], df["petal width (cm)"])
plt.title("petal length vs width")
plt.xlabel("petal length")
plt.ylabel("petal width")
```

```python
plt.boxplot(df["sepal length (cm)"])
plt.title("Sepal Length Box Plot")
```

```python
plt.figure(figsize = (12, 8))
plt.pie(df["target"].value_counts(), labels = ["Satosa", "Versicolor", "Virginica"])
plt.legend()
plt.show()
```
