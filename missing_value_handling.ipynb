```python
import pandas as pd
```

```python
df = pd.read_csv("D:/Download/heart_missing_values.csv")
print(df.head())
print(df.tail())
print(df.shape)
print(df.info())
```

```python
#1. drop missing values
df1= df.dropna()
print("after dropping missing values : ")
display(df1)
```

```python
#2. fill missing values with 0
df2 = df.fillna(0)
print("after filling with zero : ")
display(df2)
```

```python
print("fill missing values with mean : ")
df3 = df.fillna(df.mean())
display(df3)
```
