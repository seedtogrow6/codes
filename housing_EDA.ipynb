```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Dataset
df = pd.read_csv("housing.csv")

# Basic EDA
print("First 5 Rows:")
display(df.head())

print("Dataset Shape:", df.shape)

print("Column Names:")
print(df.columns)

print("Dataset Information:")
df.info()

print("Statistical Summary:")
display(df.describe())

print("Missing Values:")
print(df.isnull().sum())

print("Duplicate Rows:", df.duplicated().sum())
```

```python
# 1. Histogram
plt.figure(figsize=(6, 4))
plt.hist(df["median_house_value"], bins=20, color="skyblue", edgecolor="black")
plt.title("House Value Distribution")
plt.xlabel("Median House Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# 2. Scatter Plot
plt.figure(figsize=(6, 4))
plt.scatter(df["median_income"], df["median_house_value"], color="green", s=5)
plt.title("Income vs House Value")
plt.xlabel("Median Income")
plt.ylabel("Median House Value")
plt.show()
```

```python
# 3. Bar Chart
count = df["ocean_proximity"].value_counts()

plt.figure(figsize=(7, 4))
plt.bar(count.index, count.values, color="orange")
plt.title("Houses by Ocean Proximity")
plt.xlabel("Ocean Proximity")
plt.ylabel("Number of Records")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# 4. Box Plot
plt.figure(figsize=(6, 4))
plt.boxplot(df["median_house_value"].dropna())
plt.title("House Value Box Plot")
plt.ylabel("Median House Value")
plt.show()
```

```python
# 5. Line Plot
age_price = df.groupby("housing_median_age")["median_house_value"].mean()

plt.figure(figsize=(6, 4))
plt.plot(age_price.index, age_price.values, marker="o", markersize=3)
plt.title("Housing Age vs Average House Value")
plt.xlabel("Housing Median Age")
plt.ylabel("Average House Value")
plt.grid(True)
plt.show()
```

```python
# 6. Correlation Heatmap
corr = df.corr(numeric_only=True)

plt.figure(figsize=(9, 6))
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar(label="Correlation")

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Housing Correlation Heatmap")
plt.tight_layout()
plt.show()
```
